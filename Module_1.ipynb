```python
XTX = X.T @ X
XTX
```

```python
XTX = np.linalg.inv(XTX)
XTX
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = (XTX @ X.T) @ y
w
```
